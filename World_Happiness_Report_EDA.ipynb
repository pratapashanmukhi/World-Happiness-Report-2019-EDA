# World Happiness Report 2019 - Exploratory Data Analysis

## Objective

The objective of this project is to analyze the World Happiness Report 2019 dataset and identify factors that influence happiness across different countries.

```python


import pandas as pd
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('2019.csv')

df.head()
```

## Dataset Overview

```python
print("Shape of Dataset:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nData Types:")
print(df.dtypes)
```

### Observation

- The dataset contains information about 156 countries.
- There are 9 attributes in the dataset.
- No missing values were found.
- Most columns are numerical except the country name column.
- The dataset is clean and suitable for analysis.

```python
df.describe()
```

### Summary Statistics Observation

- The average happiness score is around 5.4.
- Happiness scores range from approximately 2.8 to 7.8.
- GDP per capita and social support values vary significantly among countries.
- Countries with higher economic indicators generally show higher happiness scores.

## Top 10 Happiest Countries

```python
top10 = df.nlargest(10,'Score')[['Country or region','Score']]
print(top10)
```

### Observation

- Finland ranks as the happiest country.
- Nordic countries dominate the top positions.
- Countries with strong social support and economic stability appear in the top rankings.

## Bottom 10 Happiest Countries

```python
bottom10 = df.nsmallest(10,'Score')[['Country or region','Score']]
print(bottom10)
```

### Observation

- South Sudan has the lowest happiness score.
- Most countries in the bottom group face economic or social challenges.
- Lower GDP and weaker social support are common among these countries.

## GDP per Capita vs Happiness Score

```python
plt.figure(figsize=(8,6))
plt.scatter(df['GDP per capita'], df['Score'])

plt.title('GDP per Capita vs Happiness Score')
plt.xlabel('GDP per Capita')
plt.ylabel('Happiness Score')

plt.show()
```

```python
correlation = df['GDP per capita'].corr(df['Score'])

print("Correlation between GDP per capita and Happiness Score:", correlation)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(df['Social support'], df['Score'])

plt.title('Social Support vs Happiness Score')
plt.xlabel('Social Support')
plt.ylabel('Happiness Score')

plt.show()
```

```python
print("Correlation:", df['Social support'].corr(df['Score']))
```

### Observation

- Social support has a strong positive relationship with happiness.
- Countries with better support systems tend to report higher happiness levels.

## Healthy Life Expectancy vs Happiness Score

```python
plt.figure(figsize=(8,6))
plt.scatter(df['Healthy life expectancy'], df['Score'])

plt.title('Healthy Life Expectancy vs Happiness Score')
plt.xlabel('Healthy Life Expectancy')
plt.ylabel('Happiness Score')

plt.show()
```

```python
print("Correlation:", df['Healthy life expectancy'].corr(df['Score']))
```

### Observation

- Countries with higher life expectancy generally show higher happiness scores.
- Good healthcare and living conditions contribute positively to happiness.

# Final Conclusion

- Finland is the happiest country in the 2019 report.
- GDP per capita, social support, and healthy life expectancy strongly influence happiness.
- Countries with stronger economies and better social systems tend to achieve higher happiness scores.
- The analysis shows that both economic and social factors play important roles in overall well-being.
